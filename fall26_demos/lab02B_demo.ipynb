### First steps in Python

Python uses building blocks that are very similar to what you find in other languages: if, then, for, while... and print!

```python
name = "Python Workshop"
print("Hello, "+ name + "!")
```

In most languages, the keyword **end** is used to tell when a loop/condition/definition ends.

Or some curly brackets.

In Python, there is no **end**. Instead, the ranges are defined according to indentation.

```python
# In most languages, indentation is optional and only for lisibility
x = 0
for i in range(5):
    x = x + i
    print("x is "+str(x))
```

```python
# In python, indentation (hence lisibility) is part of the language itself!
# The price: indentation errors translate into code errors
x = 0
for i in range(5):
    x = x + i
print("x is "+str(x))
```

When you make a mistake, you get some error message that (may?) help you.

```python
print("This line is correct")
for i in range(-3, 3):
    inv_i = 1/i
    print(f"The inverse of {i} is {inv_i}")
print("Will I get there?")
```

Defining functions is performed with **def**. 

```python
def my_division(x, y):
    return x / y
my_division(10, 2)
```

You can assign default values, which allows highly flexible function calls.

```python
def my_division(x=10, y=2):
    return x / y
print(my_division(10, 2))
print(my_division(7))
print(my_division(y=4))
print(my_division(y=3, x=21))
```

Packages can offer additional features. Many useful packages are shipped with Anaconda.

```python
import numpy as np # numerical computation package. Very useful so we shorten its name as np
from matplotlib import pyplot as plt # pyplot is a submodule (part of a module) to easily display figures
x = np.linspace(-4, 4)
plt.plot(x, np.sin(x), label='sin(x)')
plt.legend()
plt.show()
```

### Magics

Magics are used at the start of a cell or a line to change the way the code cell is interpreted. There are a lot of them by default (and a lot that can be added).

Time it... to make stats about the execution time.

```python
%timeit 6*7
```

Time it is useful for fast benchmarking.

```python
import numpy as np
%timeit x = np.sum([np.random.rand()**2 for _ in range(10000)])
```

```python
%timeit x = np.sum(np.random.rand(10000)**2)
```
